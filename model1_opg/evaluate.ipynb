# Model 1 — Panoramic X-ray (OPG) — Evaluation

Per-class Precision / Recall / F1 / mAP, confusion matrix, and a data-leakage check. Nothing is estimated — all values computed from the weights + test split.

## 1. Setup

```python
!pip -q install "numpy<2.0" "ultralytics==8.3.0"
from ultralytics import YOLO
WEIGHTS = "models/best.pt"
DATA = "data/data.yaml"
```

## 2. Per-class metrics (P / R / F1 / mAP)

```python
m = YOLO(WEIGHTS)
mt = m.val(data=DATA, split="test", imgsz=1024, device=0, plots=True)
print("OVERALL mAP50:", round(float(mt.box.map50),4), "| mAP50-95:", round(float(mt.box.map),4))
for i, idx in enumerate(mt.box.ap_class_index):
    p=float(mt.box.p[i]); r=float(mt.box.r[i]); f1=2*p*r/(p+r) if (p+r) else 0
    print(f"  {m.names[int(idx)]:12s} P={p:.3f} R={r:.3f} F1={f1:.3f} mAP50={float(mt.box.ap50[i]):.3f}")
```
