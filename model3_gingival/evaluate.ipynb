# Model 4 - Gingival Inflammation - Evaluation

Precision / Recall / F1 / mAP on the held-out test split, plus confusion matrix. All values computed from the weights + test split - nothing hand-entered.

## 1. Setup

```python
!pip -q install -U ultralytics
from ultralytics import YOLO
WEIGHTS = "models/best.pt"
DATA = "merged/data.yaml"
```

## 2. Metrics (P / R / F1 / mAP)

```python
m = YOLO(WEIGHTS)
mt = m.val(data=DATA, split="test", imgsz=640, device=0, plots=True)
p=float(mt.box.mp); r=float(mt.box.mr); f1=2*p*r/(p+r) if (p+r) else 0
print("gingivitis  P=%.3f R=%.3f F1=%.3f mAP50=%.3f mAP50-95=%.3f" % (p, r, f1, float(mt.box.map50), float(mt.box.map)))
```
